```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
```

##### Data understanding and preparation

Loading the Data

```python
energy_dataset = pd.read_csv("energy_dataset.csv")
weather_dataset = pd.read_csv("weather.csv")

#energy_dataset["time"] = pd.to_datetime(energy_dataset["time"])
#weather_dataset["time"] = pd.to_datetime(weather_dataset["time"])

energy_dataset['time'] = pd.to_datetime(energy_dataset['time'], utc=True, errors='coerce')
weather_dataset["time"] = pd.to_datetime(weather_dataset["time"], utc=True, errors='coerce')

df = pd.merge(energy_dataset, weather_dataset, on="time", how="inner")
df
```

### Data Types of the Columns

```python
df.dtypes
```

The dataset is composed almost entirely of numerical variables (**float64**), covering all energy generation measurements and weather-related data. The only non-numerical column is **time**, stored as an **object**, indicating a temporal value that should be converted to a proper datetime format.

##### Descriptive Analysis

```python
df.info
```

```python
pd.set_option('display.max_columns', None)
df.describe()
```

The descriptive analysis of the numerical variables highlights several important characteristics of the dataset:

- **Count:** Most variables present a consistent number of observations, close to *35000*. However, variables such as 'generation_hydro_pumped_storage_aggregated' and 'forecast_wind_offshore_eday_ahead' do not present any number of observations, meaning they don't display any relevant data information.

- **Mean:** There are a significant variety  of the average values across energy sources. For example, the variables 'generation_fossil_gas' and 'generation_fossil_hard_coal' show high mean outputs, while many others have means = 0, indicating no recorded production


##### Target Variable

Once the goal of the project is to predict the next hour's electricity price, the target variable is **'price_day_ahead'**

```python
target = df['price_day_ahead']
target_counts = target.value_counts()
target_percent = target.value_counts(normalize=True)*100
#percent100 = target.value_counts(normalize=True).mul(100).round(1).astype(str) + '%'
pd.DataFrame({"Count":target_counts, "Percentage": target_percent})
```

The analysis of the target variable shows that most price values occur only once or few time, indicating a highly continuous distribution with limited repetition. A small set of rounded prices (such as '40.00', '50.00', '45.00' and '60.00') appears more frequently, suggesting common market price levels. Overall, the high variability and the large number of unique values, confirm that the target behaves as a continuous variable with no strong discrete patterns

##### Data Visualization 

```python
plt.figure(figsize=(10,6))
sns.histplot(df['price_day_ahead'], bins=30, kde=True)
plt.title('Distribution of price_day_ahead')
plt.xlabel('price_day_ahead')
plt.ylabel('frequency')
plt.show()

df['price_day_ahead_bins'] = pd.cut(df['price_day_ahead'], bins=10, labels=False)

# Visualizar a distribuição dos bins
plt.figure(figsize=(10, 6))
df['price_day_ahead_bins'].value_counts().sort_index().plot(kind='bar')
plt.title('Distribution of price_day_ahead in Bins')
plt.xlabel('price_day_ahead bins')
plt.ylabel('frequency')
plt.show()
```

The distribution of *'price_day_ahead' shows a pattern with most prices concentrated between 40 and 60. The histograms comfirms that prices follow a distribution close to normality, with a smooth peak around 50 euros, meaning this is the most common price range in the dataset.

When the variable is divided into 10 bins, the central bins (especialy 4.0 and 5.0) contain the largest number of observations, which reinfornces that mid-range prices dominate the dataset. The frequencies drop in the extreme bins (0.0, 1.0, 8.0 and 9.0), indicating that very low and very high prices are rare.

Overall, these results show that electricity prices are relatively stable, concentrated around a central value.

##### Data Visualization

```python
plt.figure(figsize=(20, 25))
for i, col in enumerate(num_columns):
    plt.subplot((len(num_columns) + 2) // 3, 3, i + 1)
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f'Distribution {col}')
    plt.xlabel(col)
    plt.ylabel('frequency')

plt.tight_layout()
plt.show()
```

Ao observar a distribuição dos dados, é possível concluir a irrelevância de alguns preditores, nomeadamente: generation_fossil_coal-derived_gas, generation_fossil_oil_shale, generation_fossil_peat, generation_geothermal, generation_marine and generation_wind_offshore, uma vez que não oferecem grande informação para o dataset.

```python
df['hour'] = df['time'].dt.hour          
df['day_of_week'] = df['time'].dt.dayofweek  
df['month'] = df['time'].dt.month          

#print(df[['time', 'hour', 'day_of_week', 'month']].head())

# Preço médio por hora do dia
hourly_price = (
    df.groupby('hour', as_index=False)['price_day_ahead']
      .mean()
      .rename(columns={'price_day_ahead': 'price_mean'})
)

plt.figure(figsize=(10, 4))
sns.lineplot(data=hourly_price, x='hour', y='price_mean', marker='o')
plt.title("Average price per hour of the day")
plt.xlabel("Time of day")
plt.ylabel("Average price (day ahead)")
plt.grid(True)
plt.tight_layout()
plt.show()

# 3.2 Preço médio por dia da semana
dow_price = (
    df.groupby('day_of_week', as_index=False)['price_day_ahead']
      .mean()
      .rename(columns={'price_day_ahead': 'price_mean'})
)

plt.figure(figsize=(10, 4))
sns.lineplot(data=dow_price, x='day_of_week', y='price_mean', marker='o')
plt.title("Average price per day of the week (0 = Monday, 6 = Sunday)")
plt.xlabel("Day of week")
plt.ylabel("Average price (day ahead)")
plt.grid(True)
plt.tight_layout()
plt.show()

# 3.3 Heatmap: preço médio por hora x dia da semana
pivot = (
    df.groupby(['day_of_week', 'hour'])['price_day_ahead']
      .mean()
      .unstack()  # índice = dia_da_semana, colunas = hora
)

plt.figure(figsize=(12, 5))
sns.heatmap(pivot, annot=False)
plt.title("Average price per hour x day of the week")
plt.xlabel("Time of day")
plt.ylabel("Day of the week (0 = Mon, 6 = Sun)")
plt.tight_layout()
plt.show()
```

```python

df['hour'] = df['time'].dt.hour  
df['day'] = df['time'].dt.day        
df['day_of_week'] = df['time'].dt.dayofweek  
df['month'] = df['time'].dt.month    

plt.figure(figsize=(10, 6))
sns.boxplot(x='hour', y='price_day_ahead', data=df)
plt.title('Price variation throughout the day')
plt.xlabel('Time of day')
plt.ylabel('Price (day ahead)')
plt.show()

plt.figure(figsize=(10, 6))
sns.boxplot(x='day_of_week', y='price_day_ahead', data=df)
plt.title('Price variation throughout the week (0=Mon, 6=Sun)')
plt.xlabel('Day of week')
plt.ylabel('Price (day ahead)')
plt.show()

plt.figure(figsize=(10, 6))
sns.boxplot(x='month', y='price_day_ahead', data=df)
plt.title('Price variation throughout the months')
plt.xlabel('Month')
plt.ylabel('Price (day ahead)')
plt.show()
```

#### Correlation Matrix

```python
label_encoder = LabelEncoder()

df_encoded = df.copy()

# Aplicar a codificação para todas as colunas categóricas
for column in df_encoded.select_dtypes(include=['object', 'category']).columns:
    df_encoded[column] = label_encoder.fit_transform(df_encoded[column].astype(str))


# Calcular a matriz de correlação
correlation_matrix_all = df_encoded.corr()

# Exibir a matriz de correlação
plt.figure(figsize=(20, 15))
sns.heatmap(correlation_matrix_all, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()
```

Como se pode concluir, também, através da matriz de correlação, existem alguns preditores que não são relevantes para o dataset.

#### Statistical Models

```python
# selecionar 3000 linhas do dataset aleaoriamente
df_sample = df_encoded.sample(n=3000, random_state=42)
df_sample
```

Selecionou-se um subconjunto de 3000 amostras aleatórias e definiu-se inicialmente a frequência para 5 dias de modo a se visualizar a série graficamente

```python
df_sample = df.copy() 

df_sample = df_sample.set_index('time').sort_index()

print("Index dtype:", df_sample.index.dtype)
print("Index min:", df_sample.index.min())
print("Index max:", df_sample.index.max())

D_price_day_ahead = df_sample['price_day_ahead'].resample('5D').mean().dropna()

print("Número de pontos depois do resample:", len(D_price_day_ahead))
print(D_price_day_ahead.head())

plt.figure(figsize=(12, 6))
plt.plot(D_price_day_ahead)
plt.title('Price day ahead every 5 days')
plt.xlabel('Date')
plt.ylabel('Price day ahead')
plt.grid(True)
plt.show()
```
